```python
# Experiment 8e: inspect a candidate EXTERNAL dataset (CPU only, 2 minutes). Add the dataset as input, e.g. an OASIS-based Alzheimer MRI set.
import os, collections
from PIL import Image
import numpy as np, pandas as pd
ROOT = "/kaggle/input"
rows = []
for r, d, f in os.walk(ROOT):
    imgs = [x for x in f if x.lower().endswith((".jpg", ".jpeg", ".png", ".tif", ".tiff"))]
    if imgs: rows.append((r, len(imgs), imgs[:3]))
print("Folders containing images:", len(rows))
for r, n, ex in rows[:60]: print(f"{n:7d}  {r}   e.g. {ex}")
tot = sum(n for _, n, _ in rows); print("TOTAL images:", tot)
# sample image properties and file-name patterns (names can reveal subject / slice ids)
for r, n, ex in rows[:6]:
    p = os.path.join(r, ex[0]); im = Image.open(p); a = np.asarray(im)
    print(p, "| size", im.size, "| mode", im.mode, "| dtype", a.dtype, "| min/max", a.min(), a.max())
names = [x for r, n, ex in rows[:3] for x in os.listdir(r)[:15]]
print("First file names:", names)
```
